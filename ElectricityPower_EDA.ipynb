### Proses imputasi dilakukan

--- 

Imputasi dilakukan dengan mengambil nilai mean dari region yang sama + day of week yang sama + hour yang sama dari rentang waktu 2 tahun sebelum hari H. Selain itu, karena dataset yang tersedia berukuran besar dan untuk simplicity dalam imputasi, maka dataset 2015H2 s.d. 2017H1 dijadikan sumber imputasi, 2017H2 s.d. 2023H2 (training), dan 2024H1 s.d. 2026H2 (validasi).

```python
keep_modeling_cols = [
    "Data Date",
    "Hour Number",
    "Local Time at End of Hour",
    "UTC Time at End of Hour",
    "Demand (MW)",
    "Region"
]

all_dfs = []

for dataset_path in dataset_paths:
    df = pd.read_csv(dataset_path)

    existing_cols = [col for col in keep_modeling_cols if col in df.columns]
    temp = df[existing_cols].copy()

    temp["source_file"] = Path(dataset_path).stem

    all_dfs.append(temp)

df_merge = pd.concat(all_dfs, ignore_index=True)

# Convert timestamp
df_merge["timestamp_utc"] = pd.to_datetime(
    df_merge["UTC Time at End of Hour"],
    errors="coerce",
    utc=True
)

# Convert target to numeric
df_merge["Demand (MW)"] = pd.to_numeric(df_merge["Demand (MW)"], errors="coerce")

# Periode:
# 2015H2–2017H1 = warm-up/lookback only
# 2017H2–2026H1 = modeling period
warmup_start_date = pd.Timestamp("2015-07-01", tz="UTC")
modeling_start_date = pd.Timestamp("2017-07-01", tz="UTC")
end_date = pd.Timestamp("2026-07-01", tz="UTC")  # exclusive, sampai 2026-06-30

df_merge_clipped = df_merge[
    (df_merge["timestamp_utc"] >= warmup_start_date) &
    (df_merge["timestamp_utc"] < end_date)
].copy()

df_merge_clipped = df_merge_clipped.sort_values(
    ["Region", "timestamp_utc"]
).reset_index(drop=True)

df_merge_clipped["is_modeling_period"] = (
    df_merge_clipped["timestamp_utc"] >= modeling_start_date
).astype(int)

print("Warm-up + modeling dataset size:", df_merge_clipped.shape)
print("Time range:", df_merge_clipped["timestamp_utc"].min(), "to", df_merge_clipped["timestamp_utc"].max())
print("Modeling start:", modeling_start_date)
print("Regions:", df_merge_clipped["Region"].unique())

df_merge_clipped.head()
```

```python
def make_complete_hourly_ba(group):
    ba = group["Balancing Authority"].iloc[0]
    region = group["Region"].dropna().iloc[0] if group["Region"].notna().any() else np.nan

    group = group.sort_values("timestamp_utc").drop_duplicates("timestamp_utc")
    group = group.set_index("timestamp_utc")

    full_index = pd.date_range(
        start=group.index.min(),
        end=group.index.max(),
        freq="h",
        tz="UTC"
    )

    group = group.reindex(full_index)
    group.index.name = "timestamp_utc"

    group["Balancing Authority"] = ba
    group["Region"] = region

    return group.reset_index()


df_complete_ba = (
    df_merge_clipped
    .groupby("Balancing Authority", group_keys=False)
    .apply(make_complete_hourly_ba)
    .reset_index(drop=True)
)

df_complete_ba = df_complete_ba.sort_values(
    ["Balancing Authority", "timestamp_utc"]
).reset_index(drop=True)

print("BA-level complete hourly shape:", df_complete_ba.shape)
print("Number of BAs:", df_complete_ba["Balancing Authority"].nunique())

df_complete_ba.head()
```

```python
df_complete_ba["year"] = df_complete_ba["timestamp_utc"].dt.year
df_complete_ba["month"] = df_complete_ba["timestamp_utc"].dt.month
df_complete_ba["utc_hour"] = df_complete_ba["timestamp_utc"].dt.hour
df_complete_ba["day_of_week"] = df_complete_ba["timestamp_utc"].dt.dayofweek

# Untuk sementara pakai utc_hour sebagai hour grouping.
# Nanti untuk calendar feature final, sebaiknya pakai local time / region timezone.
df_complete_ba["hour"] = df_complete_ba["utc_hour"]

df_complete_ba.head()
```

```python
def causal_impute_demand_2y_no_fallback(
    df,
    target_col="Demand (MW)",
    region_col="Region",
    time_col="timestamp_utc",
    min_group_count=5,
    lookback_years=2
):
    df = df.copy()
    df = df.sort_values([region_col, time_col]).reset_index(drop=True)

    df["target_demand_mw"] = df[target_col]
    df["is_target_imputed"] = 0
    df["imputation_method"] = "observed"

    result_parts = []

    for region, g in df.groupby(region_col):
        g = g.copy().sort_values(time_col).reset_index(drop=True)

        missing_indices = g.index[g["target_demand_mw"].isna()]

        for idx in missing_indices:
            current_time = g.loc[idx, time_col]
            current_hour = g.loc[idx, "hour"]
            current_dow = g.loc[idx, "day_of_week"]

            lookback_start = current_time - pd.DateOffset(years=lookback_years)

            hist_group = g[
                (g[time_col] < current_time) &
                (g[time_col] >= lookback_start) &
                (g[target_col].notna()) &
                (g["hour"] == current_hour) &
                (g["day_of_week"] == current_dow)
            ]

            if len(hist_group) >= min_group_count:
                g.loc[idx, "target_demand_mw"] = hist_group[target_col].mean()
                g.loc[idx, "is_target_imputed"] = 1
                g.loc[idx, "imputation_method"] = "region_dow_hour_prev2y"
            else:
                g.loc[idx, "imputation_method"] = "not_imputed"

        result_parts.append(g)

    out = pd.concat(result_parts, ignore_index=True)
    out = out.sort_values([region_col, time_col]).reset_index(drop=True)

    return out
```

```python
warmup_start_date = pd.Timestamp("2015-07-01", tz="UTC")
modeling_start_date = pd.Timestamp("2017-07-01", tz="UTC")

train_end_date = pd.Timestamp("2024-01-01", tz="UTC")  # train sampai 2023-12-31 23:00
valid_end_date = pd.Timestamp("2025-01-01", tz="UTC")  # valid sepanjang 2024
test_end_date = pd.Timestamp("2026-05-07", tz="UTC")

threshold_results = []

for min_count in [5, 10, 15, 20, 30]:
    temp = causal_impute_demand_2y_no_fallback(
        df_complete,
        target_col="Demand (MW)",
        region_col="Region",
        time_col="timestamp_utc",
        min_group_count=min_count,
        lookback_years=2
    )

    # Validation/test tidak boleh pakai target hasil imputasi
    eval_mask = temp["timestamp_utc"] >= train_end_date
    temp.loc[eval_mask, "target_demand_mw"] = temp.loc[eval_mask, "Demand (MW)"]
    temp.loc[eval_mask, "is_target_imputed"] = 0
    temp.loc[eval_mask, "imputation_method"] = np.where(
        temp.loc[eval_mask, "Demand (MW)"].notna(),
        "observed_eval_period",
        "missing_eval_period_not_imputed"
    )

    train_mask = (
        (temp["timestamp_utc"] >= modeling_start_date) &
        (temp["timestamp_utc"] < train_end_date)
    )

    valid_mask = (
        (temp["timestamp_utc"] >= train_end_date) &
        (temp["timestamp_utc"] < valid_end_date)
    )

    test_mask = (
        (temp["timestamp_utc"] >= valid_end_date) &
        (temp["timestamp_utc"] < test_end_date)
    )

    threshold_results.append({
        "min_group_count": min_count,
        "train_remaining_missing": temp.loc[train_mask, "target_demand_mw"].isna().sum(),
        "train_imputed_rows": temp.loc[train_mask, "is_target_imputed"].sum(),
        "valid_actual_missing": temp.loc[valid_mask, "Demand (MW)"].isna().sum(),
        "test_actual_missing": temp.loc[test_mask, "Demand (MW)"].isna().sum(),
    })

threshold_results_df = pd.DataFrame(threshold_results)
threshold_results_df
```

```python
def count_missing(x):
    return x.isna().sum()

def count_notna(x):
    return x.notna().sum()

df_region_hourly = (
    df_imputed_ba
    .groupby(["Region", "timestamp_utc"], as_index=False)
    .agg(
        data_date=("data_date", "min"),

        # Target actual region demand.
        # Ini hanya reliable kalau semua BA demand actual tersedia.
        actual_region_demand_mw=("Demand (MW)", "sum"),

        # Target final untuk training/lag history.
        # Ini boleh berisi BA-level imputation pada train period.
        target_region_demand_mw=("target_demand_mw", "sum"),

        # Demand Forecast sebagai benchmark eksternal, bukan fitur.
        demand_forecast_mw=("Demand Forecast (MW)", "sum"),

        num_ba=("Balancing Authority", "nunique"),
        num_rows=("Balancing Authority", "size"),
        num_actual_missing_ba=("Demand (MW)", count_missing),
        num_actual_available_ba=("Demand (MW)", count_notna),
        num_imputed_ba=("is_target_imputed", "sum")
    )
)

# Kalau semua BA actual missing pada timestamp itu, sum bisa jadi 0.
# Kita ubah actual_region_demand_mw jadi NaN jika tidak ada actual demand sama sekali.
df_region_hourly.loc[
    df_region_hourly["num_actual_available_ba"] == 0,
    "actual_region_demand_mw"
] = np.nan

# Flag apakah actual region total reliable untuk evaluation.
# Strict rule: semua BA dalam region-hour harus punya actual Demand.
df_region_hourly["is_actual_region_complete"] = (
    df_region_hourly["num_actual_missing_ba"] == 0
).astype(int)

# Kalau target_region_demand_mw tidak punya komponen sama sekali, set NaN
# Biasanya jarang, tapi untuk safety.
df_region_hourly.loc[
    df_region_hourly["target_region_demand_mw"] == 0,
    "target_region_demand_mw"
] = np.nan

df_region_hourly = df_region_hourly.sort_values(
    ["Region", "timestamp_utc"]
).reset_index(drop=True)

print("Region-level hourly shape:", df_region_hourly.shape)
print("Regions:", sorted(df_region_hourly["Region"].unique()))
display(df_region_hourly.head())
```
