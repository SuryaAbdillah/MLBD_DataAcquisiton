## Open-Meteo Historical Weather API

### Data historis

```python
import requests
import pandas as pd

url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude": 34.05,       # contoh: Los Angeles
    "longitude": -118.24,
    "start_date": "2024-01-01",
    "end_date": "2024-01-07",
    "hourly": [
        "temperature_2m",
        "relative_humidity_2m",
        "wind_speed_10m",
        "precipitation"
    ],
    "timezone": "UTC"
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()["hourly"]
weather_df = pd.DataFrame(data)

weather_df["time"] = pd.to_datetime(weather_df["time"])
weather_df.head()
```

### forecast

```python
import requests
import pandas as pd

def fetch_open_meteo_forecast(latitude, longitude, forecast_days=2):
    url = "https://api.open-meteo.com/v1/forecast"

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": [
            "temperature_2m",
            "relative_humidity_2m",
            "wind_speed_10m",
            "precipitation"
        ],
        "forecast_days": forecast_days,
        "timezone": "UTC"
    }

    response = requests.get(url, params=params)
    response.raise_for_status()

    data = response.json()["hourly"]
    df = pd.DataFrame(data)
    df["time"] = pd.to_datetime(df["time"])

    return df
```

```python
weather_forecast = fetch_open_meteo_forecast(
    latitude=34.05,
    longitude=-118.24,
    forecast_days=2
)

weather_forecast.head()
```
